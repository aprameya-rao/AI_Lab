Vaccum Cleaner

```python
class Room:
  def __init__(self,clean):
    self.clean=clean

  def isClean(self):
    return self.clean

  def do_clean(self):
    self.clean=True

class VaccumCleaner:

  def __init__(self, position):
    self.position = position

  def move(self, direction):
    if(direction == 'right'):
      self.position = 'B'
      print("Vaccum moving to B")
    elif(direction == 'left'):
      self.position = 'A'
      print("Vaccum moving to A")

  def vaccum_cleaner(self):
    while(True):
      if(A.isClean() and B.isClean()):
        print("All rooms are clean")
        break
      if(self.position == 'A' and A.isClean()):
        self.move("right")
      elif (self.position == 'B' and B.isClean()):
        self.move("left")
      elif(self.position == 'A' and not A.isClean()):
        print("Room A cleaning")
        A.do_clean()
        self.move("right")
      elif(self.position == 'B' and not B.isClean()):
        print("Room B cleaning")
        B.do_clean()
        self.move("left")

A=Room(True)
B=Room(False)
vac=VaccumCleaner('A')
vac.vaccum_cleaner()
```

Tic Tac Toe - Human v/s Machine

```python
import random

class board:
  def __init__(self):
    self.board=[[' ',' ',' '],[' ',' ',' '],[' ',' ',' ']]
    self.player='X'
    self.machine='O'

  def player_winCheck(self):
    for i in range(3):
      if(self.board[i][0]==self.board[i][1]==self.board[i][2]==self.player):
        return True
      if(self.board[0][i]==self.board[1][i]==self.board[2][i]==self.player):
        return True
    if(self.board[0][0]==self.board[1][1]==self.board[2][2]==self.player):
      return True
    if(self.board[0][2]==self.board[1][1]==self.board[2][0]==self.player):
      return True
    return False

  def machine_winCheck(self):
    for i in range(3):
      if(self.board[i][0]==self.board[i][1]==self.board[i][2]==self.machine):
        return True
      if(self.board[0][i]==self.board[1][i]==self.board[2][i]==self.machine):
        return True
    if(self.board[0][0]==self.board[1][1]==self.board[2][2]==self.machine):
      return True
    if(self.board[0][2]==self.board[1][1]==self.board[2][0]==self.machine):
      return True
    return False

  def check_draw(self):
    for i in range(3):
      for j in range(3):
        if(self.board[i][j]==' '):
          return False
    return True

  def printBoard(self):
    print("---------")
    for i in range(3):
      print("| ", end="")
      for j in range(3):
        print(self.board[i][j] + " | ", end="")
      print("\n---------")

  def player_move(self):
    while True:
      try:
        row = int(input("Enter row (0, 1, or 2): "))
        col = int(input("Enter column (0, 1, or 2): "))
        if 0 <= row <= 2 and 0 <= col <= 2:
          if self.board[row][col] == ' ':
            self.board[row][col] = self.player
            break
          else:
            print("That spot is already taken! Try again.")
        else:
          print("Invalid indices! Please enter 0, 1, or 2.")
      except ValueError:
        print("Invalid input! Please enter integers.")

  def machine_move(self):
    # 1. Check if machine can win in this turn
    for i in range(3):
      for j in range(3):
        if self.board[i][j] == ' ':
          self.board[i][j] = self.machine
          if self.machine_winCheck():
            return
          self.board[i][j] = ' '

    # 2. Check if player is winning and block them
    for i in range(3):
      for j in range(3):
        if self.board[i][j] == ' ':
          self.board[i][j] = self.player
          if self.player_winCheck():
            self.board[i][j] = self.machine
            return
          self.board[i][j] = ' '

    # 3. Otherwise, pick a random empty cell
    empty_cells = []
    for i in range(3):
      for j in range(3):
        if self.board[i][j] == ' ':
          empty_cells.append((i, j))
    if empty_cells:
      r, c = random.choice(empty_cells)
      self.board[r][c] = self.machine

  def play(self):
    self.printBoard()
    while True:
      # Player turn
      print("\nYour Turn!")
      self.player_move()
      self.printBoard()
      if self.player_winCheck():
        print("Congratulations! You win!")
        break
      if self.check_draw():
        print("It's a draw!")
        break

      # Machine turn
      print("\nMachine's Turn...")
      self.machine_move()
      self.printBoard()
      if self.machine_winCheck():
        print("Machine wins! Better luck next time.")
        break
      if self.check_draw():
        print("It's a draw!")
        break

# Run the game
game = board()
game.play()
```

8 Puzzle Mahanttan Method

```python
import heapq

class PuzzleNode:
    def __init__(self, board, parent=None, move="", g=0):
        self.board = board
        self.parent = parent
        self.move = move
        self.g = g
        self.h = self.calculate_manhattan()
        self.f = self.g + self.h

    def __lt__(self, other):
        return self.f < other.f

    def calculate_manhattan(self):
        distance = 0
        goal_positions = {
            1: (0, 0), 2: (0, 1), 3: (0, 2),
            4: (1, 0), 5: (1, 1), 6: (1, 2),
            7: (2, 0), 8: (2, 1), 0: (2, 2)
        }
        for r in range(3):
            for c in range(3):
                val = self.board[r][c]
                if val != 0:
                    target_r, target_c = goal_positions[val]
                    distance += abs(r - target_r) + abs(c - target_c)
        return distance

    def get_zero_position(self):
        for r in range(3):
            for c in range(3):
                if self.board[r][c] == 0:
                    return r, c

    def generate_neighbors(self):
        neighbors = []
        r, c = self.get_zero_position()
        moves = {
            'Up': (r - 1, c),
            'Down': (r + 1, c),
            'Left': (r, c - 1),
            'Right': (r, c + 1)
        }
        for move_name, (nr, nc) in moves.items():
            if 0 <= nr < 3 and 0 <= nc < 3:
                new_board = [row[:] for row in self.board]
                new_board[r][c], new_board[nr][nc] = new_board[nr][nc], new_board[r][c]
                neighbors.append(PuzzleNode(new_board, self, move_name, self.g + 1))
        return neighbors

def solve_puzzle(start_board):
    goal_board = [[1, 2, 3], [8, 0, 4], [7, 6, 5]]
    start_node = PuzzleNode(start_board)

    open_set = []
    heapq.heappush(open_set, start_node)
    closed_set = set()

    while open_set:
        current_node = heapq.heappop(open_set)

        if current_node.board == goal_board:
            path = []
            curr = current_node
            while curr:
                path.append((curr.board, curr.move))
                curr = curr.parent
            return path[::-1]

        board_tuple = tuple(tuple(row) for row in current_node.board)
        if board_tuple in closed_set:
            continue
        closed_set.add(board_tuple)

        for neighbor in current_node.generate_neighbors():
            neighbor_tuple = tuple(tuple(row) for row in neighbor.board)
            if neighbor_tuple not in closed_set:
                heapq.heappush(open_set, neighbor)

    return None

def print_solution(path):
    if path is None:
        print("No solution exists!")
        return
    for idx, (board, move) in enumerate(path):
        if idx == 0:
            print("Initial Board:")
        else:
            print(f"Move {idx}: {move}")
        for row in board:
            print(row)
        print()
    print(f"Total moves to solve: {len(path) - 1}")

# 0 represents the empty space
initial_board = [
    [2, 8, 3],
    [1, 6, 4],
    [7, 0, 5]
]

print_solution(solve_puzzle(initial_board))
```

Hill Climb Algorithm

```python
import random

def get_conflicts(state):
    n = len(state)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == j - i:
                conflicts += 1
    return conflicts

def get_neighbors(state):
    neighbors = []
    n = len(state)
    for col in range(n):
        for row in range(n):
            if state[col] != row:
                neighbor = list(state)
                neighbor[col] = row
                neighbors.append(neighbor)
    return neighbors

def hill_climbing():
    while True:
        current_state = [random.randint(0, 7) for _ in range(8)]
        current_conflicts = get_conflicts(current_state)

        while True:
            neighbors = get_neighbors(current_state)
            best_neighbor = None
            best_conflicts = current_conflicts

            for neighbor in neighbors:
                neighbor_conflicts = get_conflicts(neighbor)
                if neighbor_conflicts < best_conflicts:
                    best_conflicts = neighbor_conflicts
                    best_neighbor = neighbor

            if best_conflicts >= current_conflicts:
                break

            current_state = best_neighbor
            current_conflicts = best_conflicts

        if current_conflicts == 0:
            return current_state

def print_board(state):
    for row in range(8):
        line = ""
        for col in range(8):
            if state[col] == row:
                line += "Q "
            else:
                line += ". "
        print(line)

solution = hill_climbing()
print("Solution state:", solution)
print()
print_board(solution)
```
